### 데이터 확장자 변환

```python
# 전부 jpg로 변환
from PIL import Image
import os

def convert_images_to_jpg(folder_path):
    for root, _, files in os.walk(folder_path):
        for file in files:
            if file.lower().endswith(('.png', '.jpeg')):
                src_path = os.path.join(root, file)
                dst_path = os.path.join(root, os.path.splitext(file)[0] + ".jpg")

                try:
                    # PNG → RGB (배경 투명도 제거), JPEG도 RGB로 보정
                    img = Image.open(src_path).convert("RGB")
                    img.save(dst_path, "JPEG")
                    os.remove(src_path)  # 원본 삭제
                    print(f"✅ 변환 완료: {file} → {os.path.basename(dst_path)}")
                except Exception as e:
                    print(f"⚠️ 변환 실패: {file}, 에러: {e}")

# 사용 예시: '/mnt/data/data' 아래 하위 폴더 포함 전체 이미지 변환
convert_images_to_jpg("./data")
```

### 데이터 병합

```python
# 버스전용도로
import os
import shutil

# === 사용자 설정 ===
base_path = "./data"  # 전체 데이터 폴더 경로
target_folders = ["bus", "버스전용차로"]         # 합칠 폴더 이름
output_folder = "only_bus"                      # 결과 저장 폴더 이름

# === 출력 폴더 생성 ===
output_path = os.path.join(base_path, output_folder)
os.makedirs(output_path, exist_ok=True)

# === 이미지 복사 및 이름 변경 ===
image_count = 1
for folder_name in target_folders:
    folder_path = os.path.join(base_path, folder_name)
    if not os.path.isdir(folder_path):
        print(f"[경고] 폴더가 존재하지 않음: {folder_path}")
        continue

    for file_name in sorted(os.listdir(folder_path)):
        ext = os.path.splitext(file_name)[1].lower()
        if ext == ".jpg":
            src_path = os.path.join(folder_path, file_name)
            dst_filename = f"bus_{image_count:03d}.jpg"
            dst_path = os.path.join(output_path, dst_filename)
            shutil.copy2(src_path, dst_path)
            print(f"✔ 복사됨: {dst_path}")
            image_count += 1

print(f"\n✅ 총 {image_count - 1}장의 이미지가 저장되었습니다.")
```
